```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 10))

# Box positions (x, y)
boxes = {
    "Data Collection\n(Stock + News)": (0.5, 0.9),
    "Data Preprocessing": (0.5, 0.8),
    "Sentiment Analysis\n(NLP)": (0.5, 0.7),
    "Feature Engineering\n(RSI, MACD, EMA)": (0.5, 0.6),
    "Data Integration": (0.5, 0.5),
    "Model Training\n(ML + LSTM)": (0.5, 0.4),
    "Prediction\n(Price + Direction)": (0.5, 0.3),
    "Visualization\n(Graphs & Results)": (0.5, 0.2)
}

# Draw boxes
for text, (x, y) in boxes.items():
    ax.text(x, y, text, ha='center', va='center',
            bbox=dict(boxstyle="round,pad=0.4"))

# Draw arrows
positions = list(boxes.values())
for i in range(len(positions)-1):
    x1, y1 = positions[i]
    x2, y2 = positions[i+1]
    ax.annotate('', xy=(x2, y2+0.03), xytext=(x1, y1-0.03),
                arrowprops=dict(arrowstyle="->"))

# Final settings
ax.set_title("AI-Based Stock Prediction Flowchart", fontsize=14)
ax.axis('off')

plt.tight_layout()
plt.savefig("flowchart.png", dpi=300)
plt.show()
```
